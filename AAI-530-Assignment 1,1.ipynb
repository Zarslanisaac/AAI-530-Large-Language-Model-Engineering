````python
# ============================================================
# STEP 1: IMPORT LIBRARIES, CONFIGURE MODELS, AND DEFINE
#         REUSABLE FUNCTIONS
#
# Results are displayed only. No files are saved.
# Do not repeat imports or function definitions in later cells.
# ============================================================

# ------------------------------------------------------------
# STEP 1A: IMPORT LIBRARIES
# ------------------------------------------------------------

import time
from datetime import datetime, timezone

import pandas as pd
import ollama

from IPython.display import display, Markdown


# ------------------------------------------------------------
# STEP 1B: SHARED MODEL CONFIGURATION
# ------------------------------------------------------------

PROVIDER = "Ollama"
ACCESS_TYPE = "Local runtime"

MODEL_1 = "llama3.2:3b"
MODEL_2 = "qwen2.5:3b"

SYSTEM_PROMPT = (
    "You are a helpful assistant. Give accurate, practical answers. State assumptions and avoid unsupported claims."
)

TEST_PROMPT = """
A small nonprofit wants to build a chatbot that answers questions about its public volunteer handbook. It has a limited budget, needs reliable 
answers, and wants to protect confidential information. Recommend an LLM deployment approach. Compare a commercial API, a hosted open-weight model, 
and a locally run model. Discuss cost, latency, answer quality, privacy, maintenance, and reproducibility. Give a practical recommendation and three 
risks with mitigations. Keep the response under 350 words.
"""

MODEL_OPTIONS = {
    "Llama 3.2 3B": MODEL_1,
    "Qwen2.5 3B": MODEL_2,
}


# ------------------------------------------------------------
# STEP 1C: REUSABLE MODEL-CALL FUNCTION
# ------------------------------------------------------------

def call_local_model(model, prompt):
    """Run a local Ollama model and return response metrics."""

    start = time.perf_counter()

    try:
        response = ollama.chat(
            model=model,
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": prompt},
            ],
            options={"temperature": 0.2},
        )

        elapsed = time.perf_counter() - start

        return {
            "status": "success",
            "provider": PROVIDER,
            "access_type": ACCESS_TYPE,
            "model": model,
            "response": response["message"]["content"],
            "latency_seconds": round(elapsed, 3),
            "input_tokens": response.get("prompt_eval_count"),
            "output_tokens": response.get("eval_count"),
            "error": None,
        }

    except Exception as exc:
        return {
            "status": "error",
            "provider": PROVIDER,
            "access_type": ACCESS_TYPE,
            "model": model,
            "response": None,
            "latency_seconds": round(time.perf_counter() - start, 3),
            "input_tokens": None,
            "output_tokens": None,
            "error": f"{type(exc).__name__}: {exc}",
        }


# ------------------------------------------------------------
# STEP 1D: REUSABLE DISPLAY FUNCTION
# ------------------------------------------------------------

def display_comparison(results):
    """Display model metrics and generated responses."""

    display(Markdown("# LLM Comparison Results"))

    rows = []

    for method, result in results["results"].items():
        rows.append({
            "Model": method,
            "Provider": result["provider"],
            "Model ID": result["model"],
            "Access type": result["access_type"],
            "Status": result["status"],
            "Latency (seconds)": result["latency_seconds"],
            "Input tokens": result["input_tokens"],
            "Output tokens": result["output_tokens"],
        })

    display(pd.DataFrame(rows))

    for method, result in results["results"].items():
        display(Markdown(f"## {method}"))

        if result["status"] == "success":
            display(Markdown(
                result["response"] or "_Empty response_"
            ))
        else:
            display(Markdown(
                f"**Request failed:** {result['error']}"
            ))


# ------------------------------------------------------------
# STEP 1E: DISPLAY SETUP STATUS
# ------------------------------------------------------------

print("Shared setup completed successfully.")
print("Provider:", PROVIDER)
print("Access type:", ACCESS_TYPE)
print("First model:", MODEL_1)
print("Second model:", MODEL_2)
print("Output: notebook display only")
print("File saving: disabled")
print("Ollama must be running, with both models downloaded.")
print("***********************End*************************")
````

````python
# ============================================================
# PART 2: BASIC QUERY FUNCTION
# Provider: Ollama
# Model 1: llama3.2:3b
# Model 2: qwen2.5:3b
# Output: Notebook display only
# File saving: Disabled
# ============================================================

from IPython.display import display, Markdown


display(Markdown("""
# Part 2: Basic Query Function
**Purpose:** Send the same user message to two local language models and compare their responses.
"""))


# ============================================================
# STEP 1: CREATE A REUSABLE FUNCTION
# ============================================================

display(Markdown("""
## Step 1: Create a Reusable Function

The `compare_models()` function sends one user message to both models, captures their results, and displays their responses.
"""))


def compare_models(user_message):

    # --------------------------------------------------------
    # STEP 3: ACCEPT A USER MESSAGE
    # --------------------------------------------------------

    display(Markdown("### Step 3: Accept a User Message"))

    if not isinstance(user_message, str) or not user_message.strip():
        raise ValueError("Please enter a non-empty user message.")

    print("User message:", user_message)

    # --------------------------------------------------------
    # STEP 4: SEND THE MESSAGE TO BOTH MODEL ACCESS METHODS
    # --------------------------------------------------------

    display(Markdown("""
    ### Step 4: Send the Message to Both Model Access Methods

    The same message is sent to:
    1. Ollama — llama3.2:3b
    2. Ollama — qwen2.5:3b
    """))

    models = [
        ("Llama 3.2 3B", MODEL_1),
        ("Qwen2.5 3B", MODEL_2)
    ]

    results = {}

    for label, model_name in models:

        print(f"Sending message to {label} ({model_name})...")

        try:
            response = call_local_model(model_name, user_message)

            # ------------------------------------------------
            # STEP 5: CAPTURE THE RESPONSES
            # ------------------------------------------------

            results[label] = {
                "model": model_name,
                "status": response.get("status", "unknown"),
                "response": response.get("response", ""),
                "error": response.get("error", "")
            }

        except Exception as exc:
            results[label] = {
                "model": model_name,
                "status": "failed",
                "response": "",
                "error": f"{type(exc).__name__}: {exc}"
            }

    display(Markdown("""
    ### Step 5: Capture the Responses

    Each model's response, model identifier, and status
    are stored in the `results` dictionary.
    """))

    # Display the captured responses
    for label, result in results.items():

        display(Markdown(f"#### {label}"))
        print("Model:", result["model"])
        print("Status:", result["status"])

        if result["status"] == "success":
            display(Markdown(
                result["response"] or "_No response text returned._"
            ))
        else:
            print("Error:", result["error"] or "Unknown error")

    return results


# ============================================================
# STEP 2: MAKE THE USER MESSAGE EASY TO CHANGE
# ============================================================

display(Markdown("""
## Step 2: Change the User Message and Rerun the Comparison

Edit `USER_MESSAGE` below and rerun this cell to compare the models using a different question.
"""))

USER_MESSAGE = (
    "What are three benefits of artificial intelligence in education? Give an example of each."
)

print("Prompt selected:", USER_MESSAGE)


# ============================================================
# RUN THE REUSABLE FUNCTION
# ============================================================

display(Markdown("""
## Run the Comparison

The following function call sends the selected message
to both models.
"""))

comparison_results = compare_models(USER_MESSAGE)


# ============================================================
# STEP 6: Q3 — DIFFERENCES IN THE SETUP PROCESS
# ============================================================

display(Markdown("""
## Step 6: Q3 — Differences in the Setup Process

**Question:** What differences did you notice in the setup process for the two access methods?

**Answer:**

Both models use the same provider, Ollama, and run locally. Therefore, the overall setup process was similar, and I did not need separate hosted API 
keys. The main difference was that each model had to be downloaded and verified separately. I also needed to use the correct model name for each 
model and test both connections independently.

Once both models were available, I could reuse the same Python function to send the same prompt to each model. This made it easier to compare their 
responses without writing separate connection workflows.

Overall, using the same provider simplified the setup. The main requirements were ensuring Ollama was running, downloading both models, and verifying 
that each model could respond successfully.
"""))

display(Markdown("""
---
**Completed:** Part 2: Basic Query Function

- Step 1: Reusable function created.
- Step 2: User message can be changed and rerun.
- Step 3: User message accepted.
- Step 4: Message sent to both models.
- Step 5: Responses captured.
- Step 6: Q3 answered.

**Output:** Notebook display only.  
**File saving:** Disabled.
"""))
print("***********************End*************************")
````
